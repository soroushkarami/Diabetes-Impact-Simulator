```python
import json
import joblib
import sys
from pathlib import Path
```

```python
BASE_DIR = Path.cwd().parents[0]    # 'Trained Model' folder
ROOT = BASE_DIR.parent    # 'Diabetes Impact Simulator' folder

sys.path.append(str(ROOT/'diabetes_simulator'))


pipeline = joblib.load('diabetes_pipeline.pkl')
features = joblib.load('feature_names.pkl')
```

```python
w = pipeline.named_steps['model'].coef_[0]     # the Coefficients(w) of the model 
b = pipeline.named_steps['model'].intercept_   # the intercept(b) of the model
m = pipeline.named_steps['scaler'].mean_       # the scaling means
dev = pipeline.named_steps['scaler'].scale_    # the scaling standard deviation


the_params = {
    'w': w.tolist(),        # they are all np arrays so we convert them to py lists 
    'b': b.tolist(),
    'm': m.tolist(),
    'dev': dev.tolist(),
    'features': features
}


with open('model_params.json', 'w', encoding='UTF-8') as f:
    json.dump(the_params, f, indent=4)

print('The parameters extracted from the pipeline')
```
